````python
from rdflib import FOAF
````

````python
from rdflib import Graph, Namespace, Literal
g = Graph()
````

````python
fullName = Literal("John Smith")
EX = Namespace("http://example.org/")
johnURI = EX.JohnSmith
````

````python
VCARD = Namespace("http://www.w3.org/2001/vcard-rdf/3.0#")

resource = (johnURI, VCARD.FN, fullName)

print(resource)

g.add(resource)
````

````python
g.add((EX.JaneSmith, VCARD.FN, Literal("Jane Smith")))
g.add((EX.JaneSmith, VCARD.EMAIL, Literal("jsmith@example.org")))
````

````python
g.add((EX.JohnSmith, FOAF.knows, EX.JaneSmith))
````

````python
print(g.serialize(format="ttl"))
````

````python
g.namespace_manager.bind('ex', EX)
print(g.serialize(format="ttl"))
````

Spanish: Crear el recurso Sarah Jones con su nombre completo (VCARD.FN), nombre de pila (VCARD.Given) y apellido (VCARD.Family)

English: Create a resource for Sarah Jones with her full name (VCARD.FN), given name (VCARD.Given) and family name (VCARD.Family)


````python
from rdflib import Graph, URIRef, Literal, Namespace
VCARD = Namespace("http://www.w3.org/2001/vcard-rdf/3.0#")
Sarah_resource = URIRef("hhtp://example.org/SarahJones")
g.add((Sarah_resource, VCARD.FN, Literal("Sarah Jones")))
g.add((Sarah_resource, VCARD.Given, Literal("Sarah")))
g.add((Sarah_resource, VCARD.Family, Literal("Jones")))
````

Spanish: Añadir el email de John (jsmith@example.org), que aún no lo tiene

English: Add John's email address (jsmith@example.org), which he does not have yet

````python
John_resource = URIRef("http://example.org/John")
g.add((John_resource, VCARD.EMAIL, Literal("jsmith@example.org")))
````
